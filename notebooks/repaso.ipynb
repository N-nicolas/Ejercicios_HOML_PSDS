```python
# df.duplicated() Detecta las filas repetidas
df_clean = df.drop(df[df.duplicated()].index)
```

```python
df_clean.shape
```

```python
df_clean.isna().sum().sort_values(ascending=False)
```

```python
df_clean = df_clean.dropna(subset=['founded_year'])
```

```python
df_clean.shape
```

```python
df_clean['revenue_msek'] = df_clean['revenue_msek'].fillna(df_clean['revenue_msek'].median())
```

```python
df_clean.revenue_msek.isna().sum()
```

```python
industries_ls = df_clean[(df_clean.industry == 'Life Science') & (df_clean.employees >= 100)][['name', 'employees', 'revenue_msek']]

industries_ls
```

```python
industries_ls.shape[0]
```

```python
df_clean['revenue_per_employee'] = df_clean.revenue_msek/df_clean.employees
df_clean.revenue_per_employee
```

```python
df_clean['size_category'] = df_clean.employees.apply(lambda x: 'small' if x < 50 else ('medium' if 50 <= x <= 500 else 'large'))

df_clean.size_category.value_counts()
```
